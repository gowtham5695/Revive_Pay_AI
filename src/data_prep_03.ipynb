```python
import pandas as pd

df=pd.read_csv('../Data/payments_data_prepared.csv')
print(df.shape)
df.head()
```

```python
threshold=df['transaction_amount'].median()
print(threshold)
```

```python
df['is_high_value_transaction']=(df['transaction_amount']>threshold).astype(int)
df['is_high_value_transaction'].value_counts()
```

```python
df['is_repeat_failure']=(df['retry_count']>1).astype(int)
df['is_repeat_failure'].value_counts()
```

```python
df.info()
```

```python
df_encoded=pd.get_dummies(df,columns=['failure_reason','payment_method','customer_segment'])
print(df_encoded.columns.tolist())
```

```python
df_encoded['payment_recovered']=df_encoded['payment_recovered'].map({'Yes':1,'No':0})
df_encoded['payment_recovered'].value_counts()
```

```python
df_encoded.columns
df_encoded.head()
```

```python
y = df_encoded['payment_recovered']
X = df_encoded.drop(columns=['payment_recovered', 'customerID'])
print(X.shape)
```
